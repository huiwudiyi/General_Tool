```python
import pandas as pd
```

```python
# 1. 读取 CSV 文件
df = pd.read_csv("output/wenchuang_ngram_stats/token_stats.csv")
df.sort_values(by="query_ratio_0_in_0_1", ascending=False, inplace=True)
df["query_ratio_0_in_0_2"] = df.apply(lambda row: row["query_count_0"] / (1+ row["query_count_0"] + row["query_count_2"]),axis=1)
df["length"] = df.apply(lambda row: len(row["token"]),axis=1)
df = df[(df["query_count_0"] > 10) & (df["length"] > 1) & (df["query_ratio_0_in_0_1"] > 0.5) & (df["query_ratio_0_in_0_2"] > 0.8)]
bad_tokens = df['token'].tolist()
df.head(20)
```

```python
result_keys = df.groupby("pos")["token"].agg(list)
key_word_result = []
for collumn in result_keys.keys():
    res = result_keys[collumn]
    print(collumn, res)
    key_word_result += res
print(key_word_result)

# keyword 进行扩招
```

```python
## 总结规律如下：
import re
def re_judge(regexs, string):
    for pattern in regexs: #["\\d{2,4}", "[零一二三四五六七八九十百千]+首"]:
        result = re.search(pattern, string)
        if result:
            return True
    return False


def key_judge(key_word_result, string):
    for key in key_word_result:
        if key in string:
            return True
    return False

key_judge(key_word_result, "中国共党") or re_judge(["第\\d", "[一二三四五六七八九]年级|\\d年级", "第[一二三四五六七八九]|[一二三四五六七八九]名|[一二三四五六七八九]个", "[一二三四五六七八九]年级|高中生|高中学生|小学生|初中生", "初[一二三四五六七八九]"], "3年级")
```

```python
df_ngram = pd.read_csv("output/wenchuang_ngram_stats/ngram_stats.csv")
df_ngram.sort_values(by="query_ratio_0_in_0_1", ascending=False, inplace=True)
df_ngram["query_ratio_0_in_0_2"] = df_ngram.apply(lambda row: row["query_count_0"] / (1+ row["query_count_0"] + row["query_count_2"]),axis=1)
df_ngram["length"] = df_ngram.apply(lambda row: len(row["ngram"]),axis=1)
df_ngram["1gram"] = df_ngram['tokens'].apply(lambda x: False if x in bad_tokens else True)
df_ngram = df_ngram[(df_ngram["query_count_0"] > 3) & (df_ngram["length"] > 1) & (df_ngram["query_ratio_0_in_0_1"] > 0.7) & (df_ngram["query_ratio_0_in_0_2"] > 0.8)]
df_ngram.head(50)
```

```python
result_keys = df_ngram.groupby("pos")["ngram"].agg(list)
word_ngram_key_result = []
for collumn in result_keys.keys():
    res = result_keys[collumn]
    res = [r for r in res if not(key_judge(key_word_result, r) or re_judge(["第\\d", "[一二三四五六七八九]年级|\\d年级", "第[一二三四五六七八九]|[一二三四五六七八九]名|[一二三四五六七八九]个", "[一二三四五六七八九]年级|高中生|高中学生|小学生|初中生", "初[一二三四五六七八九]"], r))]
    if len(res) == 0:
        continue
    print(collumn, res)
    word_ngram_key_result += res
print(word_ngram_key_result)
```
