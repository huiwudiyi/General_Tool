```python
import json
import pandas as pd
```

```python
dataframe = pd.read_excel("资源号详情.xlsx")
dataframe.head()
```

```python
tool_description = []
for indx, row in dataframe.iterrows():
    idx = row["资源号"]
    title = row["名称"]
    description = row["描述"].replace("资源号描述：\n", "")
    tool_description.append({
        "id": idx,
        "title": title,
        "description": description,
    })
```
