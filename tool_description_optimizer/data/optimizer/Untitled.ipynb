```python
import json
import pandas as pd
```

```python
dataframe = pd.read_excel("/home/work/zhuzerun/project/aiapi_project/data/api_description/资源号详情（百看）.xlsx")

tool_descriptions = {}
for indx, row in dataframe.iterrows():
    data = {}
    data["id"] = str(row["\n"])
    if len(str(row["描述"])) < 50:
        continue
    name = row["名称"]
    if pd.isna(row["名称"]):
        name = row["垂类（重点标红）"]
    data["title"] = str(row["名称"])
    data["description"] = row["描述"]
    tool_descriptions[str(row["\n"])] = data
with open("tool_descriptions.json", "w") as w:
    json.dump(tool_descriptions, w, ensure_ascii=False, indent=2)
dataframe.head()
```
