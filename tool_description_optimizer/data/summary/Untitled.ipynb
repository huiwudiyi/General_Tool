```python
# def safe_json_dumps(obj: Any) -> str:
#     return json.dumps(obj, ensure_ascii=False, indent=2)


with open("query.json", 'w') as w:
    json.dump(query_dict, w, ensure_ascii=False, indent=2)

with open("tool_description.json", 'w') as w:
    json.dump(tool_descriptions, w, ensure_ascii=False, indent=2)
```
