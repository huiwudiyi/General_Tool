````python
def generate_string_md5(text_data: str) -> str:
    """
    生成给定字符串的 MD5 哈希值。
    """
    # 1. 字符串必须先编码为字节 (bytes)
    encoded_data = text_data.encode('utf-8')
    
    # 2. 创建 MD5 哈希对象并更新数据
    hasher = hashlib.md5(encoded_data)
    
    # 3. 获取十六进制格式的摘要
    md5_hash = hasher.hexdigest()
    
    return md5_hash

def gen_data_function(row, call_query = [], description_opt=""):
    srcid = row["resource_id"]
    name = row["names"]
    description = row["descriptions"]
    if len(description_opt) != 0:
        description = description_opt
    if len(description_opt) != 0:
        description = description_opt
    return {
	"description": name + "_aidpi server",
	"loc": "base_"+ str(int(srcid)),
    "resource_id":str(int(srcid)),
	"name": name,
#     "name": "",
	"@update": int(time.time()),
	"data_type": "aiapi",
	"id": generate_string_md5(name),
	"pos_instance": json.dumps([description.replace('"', " ").replace('\n', " ")] + call_query[:70], ensure_ascii=False),
#     "pos_instance": json.dumps([description.replace('"', " ").replace('\n', " ")] + call_query, ensure_ascii=False),
	"build_data": {
		"id": generate_string_md5(name),
		"name": name,
		"name_en": "aiapi-server_" + "".join(lazy_pinyin(name)),
		"tools": [{
			"annotations": {},
			"description": description,
			"inputSchema": {
            	"properties": {
					"question": {
						"type": "string",
						"description": "用户query"
					}
				},
				"required": ["question"],
				"type": "object"
			},
			"name": "".join(lazy_pinyin(name)) + "_" + str(srcid)
		}],
		"url": "",
		"description": name + "_aidpi server"
	},
	"channel_control": "",
	"del_flag": "0"
}

def list_file(folder_path):
    all_items = os.listdir(folder_path)
    # 只列出文件（不包括文件夹）
    files_only = [os.path.join(folder_path, item) for item in all_items if os.path.isfile(os.path.join(folder_path, item)) and item.endswith(".pkl")]
    print("\n只列出文件:")
    return files_only
# /root/.jupyter/lab/workspaces/env_run/output/zacharychu/code/tool_description_optimizer/data/data_left/summary_output/1577.pkl
files = list_file("../data/data_left/summary_output/")
files
````

````python
!unset http_proxy
!unset https_proxy
dag._load_group(state)
max_k = max(dag.k_list)
dag._sample_topk: Dict[int, Dict[str, List[str]]] = {}

for sample in dag.group_samples:
    if sample.rejected:
        print(f"[retriever] sample={sample.sample_index} 已被裁判拦截，跳过检索评估")
        dag._sample_topk[sample.sample_index] = {}
        continue
    dag._write_description(sample.description)
    evaluator = dag.rebuild_index()
    summary_df, detail_df = evaluator.evaluate(k_list=dag.k_list, eval_views=[dag.eval_view])

    row = summary_df[(summary_df["view"] == dag.eval_view) & (summary_df["k"] == max_k)]
    sample.retrieval_reward = float(row.iloc[0]["mrr_at_k"]) if len(row) else 0.0

    # 缓存每个 query 的 topk 候选，供 Selector 复用，避免重复检索
    topk_map: Dict[str, List[str]] = {}
    subset = detail_df[(detail_df["view"] == dag.eval_view) & (detail_df["k"] == max_k)]
    for _, detail_row in subset.iterrows():
        topk_map[str(detail_row["query"])] = list(detail_row["recall_ids"])
    dag._sample_topk[sample.sample_index] = topk_map

    # summary 的 pass_at_k 分母含负例（负例 gold 为空、永不 pass），会被系统性稀释，
    # 因此这里从明细里只按正例 query 求均值；MRR 本身已排除空 gold，可直接用
    sample.detail["retrieval"] = {
        "mrr_at_%d" % max_k: sample.retrieval_reward,
        "pass_at_%d" % max_k: dag._positive_pass_at_k(subset),
    }

    print(f"[retriever] sample={sample.sample_index} mrr@{max_k}={sample.retrieval_reward:.4f}")
````

````python
eval_queries = dict(dag.query_good_dict)
for query in dag.negative_queries:
    eval_queries.setdefault(query, [])

evaluator = ToolPassAtKRecallEvaluator(
    tools=dag.tools_dict,
    query_gold_ids=eval_queries,
    embeding_term=dag.embeding_term,
    encode_fn=encode_fn,
    batch_size=8,
)
evaluator.build_vector_index()
````
